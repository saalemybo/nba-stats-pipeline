```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

# Connect to the local SQLite database
conn = sqlite3.connect("../database/nba_stats.db")
```

```python
# SQL query to get Players's game-by-game stats
query = """
SELECT game_date, pts, ast, reb
FROM game_log
ORDER BY game_date;
"""

df = pd.read_sql_query(query, conn)
conn.close()

# Preview the data
df.head()
```

```python
# Convert game_date to datetime
df["game_date"] = pd.to_datetime(df["game_date"])

# Plot the data
plt.figure(figsize=(12, 6))
plt.plot(df["game_date"], df["pts"], marker="o", label="Points")
plt.plot(df["game_date"], df["ast"], marker="s", label="Assists")
plt.plot(df["game_date"], df["reb"], marker="^", label="Rebounds")

plt.title("Player - Game-by-Game Performance (2023-24)")
plt.xlabel("Game Date")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
